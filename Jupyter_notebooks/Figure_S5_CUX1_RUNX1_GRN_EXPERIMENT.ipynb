# Read variant

```r
 #### READ and transform selected_variants ----
  
  selected_variants = unlist(strsplit("rs139141690__chr7_101856650_G_A", split=","))
  
  cat("selected_variants_0\n")
  cat(sprintf(as.character(selected_variants)))
  cat("\n")
  
  rs_vector<-gsub("__.+$","",selected_variants)
  VAR_vector<-gsub("[^__]+__","",selected_variants)
  
  cat("rs_vector_0\n")
  cat(str(rs_vector))
  cat("\n")
  
  cat("VAR_vector_0\n")
  cat(str(VAR_vector))
  cat("\n")
  
  
  selected_variants_df<-as.data.frame(cbind(rs_vector,VAR_vector))
  
  colnames(selected_variants_df)<-c('rs','VAR')
  
  selected_variants_df$chr<-gsub("_.+$","",selected_variants_df$VAR)
  selected_variants_df$pos38<-gsub("^[^_]+_","",selected_variants_df$VAR)
  selected_variants_df$pos38<-as.integer(gsub("_.+$","",selected_variants_df$pos38))
  selected_variants_df$ref<-gsub("^[^_]+_[^_]+_","",selected_variants_df$VAR)
  selected_variants_df$ref<-gsub("_.+$","",selected_variants_df$ref)
  selected_variants_df$alt<-gsub("^[^_]+_[^_]+_[^_]+_","",selected_variants_df$VAR)
  
  cat("selected_variants_df_0\n")
  cat(str(selected_variants_df))
  cat("\n")
  
  gr_VARS <- GRanges(
    seqnames = as.character(gsub("chr","",selected_variants_df$chr)),
    name2=as.character(selected_variants_df$rs),
    ranges=IRanges(
      start=as.numeric(selected_variants_df$pos38),
      end=as.numeric(selected_variants_df$pos38),
      name=selected_variants_df$VAR))
  
  cat("gr_VARS_0\n")
  cat(str(gr_VARS))
  cat("\n")
```

# Read in GRN peaks

```r
  
GRN_peaks<-read.table(file="/group/soranzo/manuel.tardaguila/2025_K562_multiome_reanalysis/Downstream_analysis/simba/CUX1_RUNX1_peaks_and_genes.tsv", sep="\t", header=TRUE)

cat("GRN_peaks_0\n")
cat(str(GRN_peaks))
cat("\n")

GRN_peaks$Peak_ID<-gsub("_","-",GRN_peaks$Peak_ID)

cat("GRN_peaks_0\n")
cat(str(GRN_peaks))
cat("\n")
```

## Subset peaks with CUX1 motif

```r
GRN_peaks_CUX1<-GRN_peaks[which(GRN_peaks$CUX1 > 0 | GRN_peaks$`CUX1.1` >0),]

cat("GRN_peaks_CUX1_0\n")
cat(str(GRN_peaks_CUX1))
cat("\n")

GRN_peaks_CUX1_subset<-unique(GRN_peaks_CUX1[,which(colnames(GRN_peaks_CUX1)%in%c('Peak_ID','Gene_Name'))])

GRN_peaks_CUX1_subset$Motif<-'M_CUX1'

cat("GRN_peaks_CUX1_subset_0\n")
cat(str(GRN_peaks_CUX1_subset))
cat("\n")

```

## Subset peaks with RUNX1 motif

```r
GRN_peaks_RUNX1<-GRN_peaks[which(GRN_peaks$RUNX1 > 0),]

cat("GRN_peaks_RUNX1_0\n")
cat(str(GRN_peaks_RUNX1))
cat("\n")

GRN_peaks_RUNX1_subset<-unique(GRN_peaks_RUNX1[,which(colnames(GRN_peaks_RUNX1)%in%c('Peak_ID','Gene_Name'))])

GRN_peaks_RUNX1_subset$Motif<-'M_RUNX1'

cat("GRN_peaks_RUNX1_subset_0\n")
cat(str(GRN_peaks_RUNX1_subset))
cat("\n")

```

## Integrate

```r
GRN_peaks_DEF<-rbind(GRN_peaks_RUNX1_subset,GRN_peaks_CUX1_subset)

colnames(GRN_peaks_DEF)[which(colnames(GRN_peaks_DEF) == 'Gene_Name')]<-'gene'

GRN_peaks_DEF$Motif<-factor(GRN_peaks_DEF$Motif,
                           levels=c('M_CUX1','M_RUNX1'),
                           ordered=TRUE)

cat("GRN_peaks_DEF_0\n")
cat(str(GRN_peaks_DEF))
cat("\n")
```

## Merge with REP from DE

```r
GRN_peaks_DEF<-merge(GRN_peaks_DEF,
                     REP_genes,
                     by='gene')

cat("GRN_peaks_DEF_1\n")
cat(str(GRN_peaks_DEF))
cat("\n")
```

# Tile plot

```r
DEBUG<-1

order_from_REP<-REP_genes$gene
        
        
cat("order_from_REP_0\n")
cat(str(order_from_REP))
cat("\n")
```

```r
tile_plot<-merge(DA_results_sel,
                 GRN_peaks_DEF,
                 by="Peak_ID")


if(DEBUG == 1){
  cat("tile_plot_0\n")
  cat(str(tile_plot))
  cat("\n")
  cat(str(unique(tile_plot$Peak_ID)))
  cat("\n")
}

tile_plot$gene<-factor(tile_plot$gene, levels=rev(order_from_REP), ordered=T)



tile_plot$SIG<-NA

tile_plot$SIG[which(tile_plot$minuslog10padj >= 1.3)]<-'YES'
tile_plot$SIG[which(tile_plot$minuslog10padj < 1.3)]<-'NO'


tile_plot$SIG<-factor(tile_plot$SIG, levels=c('NO','YES'), ordered=T)

if(DEBUG == 1){
  cat("tile_plot_1\n")
  cat(str(tile_plot))
  cat("\n")
}
```

## Order the Peaks

```r
tile_plot<-tile_plot[order(tile_plot$gene),]

if(DEBUG == 1){
  cat("tile_plot_3\n")
  cat(str(tile_plot))
  cat("\n")
  cat(sprintf(as.character(unique(tile_plot$Peak_ID))))
  cat("\n")
}

order_peaks<-rev(unique(tile_plot$Peak_ID))


tile_plot$Peak_ID<-factor(tile_plot$Peak_ID,
                         levels=order_peaks,
                         ordered=T)

if(DEBUG == 1){
  cat("tile_plot_4\n")
  cat(str(tile_plot))
  cat("\n")
  cat(sprintf(as.character(unique(tile_plot$Peak_ID))))
  cat("\n")
}

tile_plot_processed <- tile_plot %>%
     mutate(myaxis = paste0(gene," ", Peak_ID)) %>%
     mutate(myaxis=fct_reorder(myaxis,as.numeric(gene)))

if(DEBUG == 1){
  cat("tile_plot_processed_0\n")
  cat(str(tile_plot_processed))
  cat("\n")
  cat(sprintf(as.character(unique(tile_plot_processed$Peak_ID))))
  cat("\n")
}
```

## keep concordant GRN motif peaks

```r
indx_CUX1<-which(tile_plot_processed$Gene_class == 'GRN_CUX1' & tile_plot_processed$Motif == 'M_CUX1')

str(indx_CUX1)
```

```r
indx_RUNX1<-which(tile_plot_processed$Gene_class == 'GRN_RUNX1' & tile_plot_processed$Motif == 'M_RUNX1')

str(indx_RUNX1)
```

```r
indx.overlap<-which(tile_plot_processed$Gene_class == 'GRN_CUX1|GRN_RUNX1')

str(indx.overlap)
```

```r
tile_plot_processed_concordant<-tile_plot_processed[c(indx_CUX1,indx_RUNX1,indx.overlap),]


cat("tile_plot_processed_concordant_0\n")
cat(str(tile_plot_processed_concordant))
cat("\n")
```

## Subset for peaks with abslogfc >= 0.05

```r
display_peaks<-tile_plot_processed_concordant$Peak_ID[which(tile_plot_processed_concordant$abslogfc >= 0.05)]

cat("display_peaks_0\n")
cat(str(display_peaks))
cat("\n")
```
