# Read Link peaks results

```r
Linked_peaks_to_DE_genes<-readRDS(file="Linked_peaks_to_DE_genes.rds")

cat("Linked_peaks_to_DE_genes_0\n")
cat(str(Linked_peaks_to_DE_genes))
cat("\n")

Linked_peaks_to_DE_genes$minuslog10padj<--1*log10(Linked_peaks_to_DE_genes$pvalue)

cat("Linked_peaks_to_DE_genes_1\n")
cat(str(Linked_peaks_to_DE_genes))
cat("\n")

Linked_peaks_to_DE_genes_sub<-unique(Linked_peaks_to_DE_genes[,which(colnames(Linked_peaks_to_DE_genes)%in%c('Peak_ID','gene','zscore','minuslog10padj'))])


cat("Linked_peaks_to_DE_genes_sub_0\n")
cat(str(Linked_peaks_to_DE_genes_sub))
cat("\n")

colnames(Linked_peaks_to_DE_genes_sub)[which(colnames(Linked_peaks_to_DE_genes_sub) == 'zscore')]<-'Link_z_score'
colnames(Linked_peaks_to_DE_genes_sub)[which(colnames(Linked_peaks_to_DE_genes_sub) == 'minuslog10padj')]<-'Link_minuslog10padj'


cat("Linked_peaks_to_DE_genes_sub_1\n")
cat(str(Linked_peaks_to_DE_genes_sub))
cat("\n")
```

# Read DA results

```r
setwd("/group/soranzo/manuel.tardaguila/2025_K562_multiome_reanalysis/Downstream_analysis/DA_per_cluster/")
```

```r
DA_results<-readRDS(file="DA_results_Diff_K562.rds")

cat("DA_results_0\n")
cat(str(DA_results))
cat("\n")

DA_results$chr<-gsub("-.+$","",DA_results$Peak_ID)
DA_results$start<-gsub("^[^-]+-","",DA_results$Peak_ID)
DA_results$start<-as.integer(gsub("-.+$","",DA_results$start))

DA_results$end<-as.integer(gsub("^[^-]+-[^-]+-","",DA_results$Peak_ID))

cat("DA_results_1\n")
cat(str(DA_results))
cat("\n")
```
