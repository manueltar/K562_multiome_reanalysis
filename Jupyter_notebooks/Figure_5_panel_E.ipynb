```r

Locus_df<-data.frame()

for(i in 1:length(selected_genes_df$gene)){

    gene_sel<-selected_genes_df$gene[i]

    cat("--------------------->\t")
    cat(sprintf(as.character(gene_sel)))
    cat("\t")
                               
   

    Linked_peaks_subset_sel<-Linked_peaks_subset[which(Linked_peaks_subset$gene == gene_sel),]
    
        if(dim(Linked_peaks_subset_sel)[1] >0){
    
            
            
       Gather_links<- data.frame(matrix(vector(), dim(Linked_peaks_subset_sel)[1], 5,
                                   dimnames=list(c(),
                                                 c("Locus_ID","Feature","chr","start","end"))),stringsAsFactors=F)

             
    
    
            Gather_links$Locus_ID<-paste("Locus", gene_sel, sep="_")   
            Gather_links$Feature<-'link'        
            Gather_links$chr<-as.character(Linked_peaks_subset_sel$seqnames)
            Gather_links$start<-as.integer(Linked_peaks_subset_sel$start)
            Gather_links$end<-as.integer(Linked_peaks_subset_sel$end)

        

            if(DEBUG == 1){
            cat("Linked_peaks_subset_sel\n")
            #cat(str(Linked_peaks_subset_sel))
            cat("\n")

            cat("Gather_links\n")
            cat(str(Gather_links))
            cat("\n")
            }
            
        }else{

            cat("No links\n")
            Gather_links<-data.frame()
        }#dim(Linked_peaks_subset_sel)[1] >0

    ensembl_gtf_gene_sel<-ensembl_gtf_gene[which(ensembl_gtf_gene$gene_name == gene_sel),]    

    if(dim(ensembl_gtf_gene_sel)[1] >0){

            Gather_gene<- data.frame(matrix(vector(), dim(ensembl_gtf_gene_sel)[1], 5,
                                   dimnames=list(c(),
                                                 c("Locus_ID","Feature","chr","start","end"))),stringsAsFactors=F)

             
    
    
            Gather_gene$Locus_ID<-paste("Locus", gene_sel, sep="_")   
            Gather_gene$Feature<-'gene'        
            Gather_gene$chr<-paste('chr',as.character(ensembl_gtf_gene_sel$seqid),sep='')
            Gather_gene$start<-as.integer(ensembl_gtf_gene_sel$start)
            Gather_gene$end<-as.integer(ensembl_gtf_gene_sel$end)

        

            if(DEBUG == 1){
            cat("ensembl_gtf_gene_sel\n")
            #cat(str(ensembl_gtf_gene_sel))
            cat("\n")

            cat("Gather_gene\n")
            cat(str(Gather_gene))
            cat("\n")
            }
            
        }else{

            cat("No genes\n")
            Gather_gene<-data.frame()
        }#dim(ensembl_gtf_gene_sel)[1] >0


    genes_and_links<-rbind(Gather_gene,Gather_links)

    if(dim(genes_and_links)[1] > 0)
    {
      if(DEBUG == 1){
                cat("genes_and_links_0\n")
                cat(str(genes_and_links))
                cat("\n")
                }

        number_of_chromosomes<-length(unique(genes_and_links$chr))

        if(number_of_chromosomes >1){

            stop("More than once chromosome in the locus\n")


        }else{


             genes_and_links$Locus_start<-as.integer(min(genes_and_links$start)-distance_offset)

            if(unique(genes_and_links$Locus_start) < 0){
    
                genes_and_links$Locus_start<-min(genes_and_links$start)
    
            }# genes_and_links$Locus_start < 0
    
            genes_and_links$Locus_end<-as.integer(max(genes_and_links$end)+distance_offset)
    
             if(DEBUG == 1){
                    cat("genes_and_links_1\n")
                    cat(str(genes_and_links))
                    cat("\n")
                    }

            Locus_df<-rbind(unique(genes_and_links[,which(colnames(genes_and_links)%in%c('Locus_ID','chr','Locus_start','Locus_end'))]),Locus_df)

        }#number_of_chromosomes >1       

    }#dim(genes_and_links)[1] > 0 

}#i in 1:length(selected_genes_df$gene


cat("Locus_df\n")
cat(str(Locus_df))
cat("\n")

 
```
